```python
from shapely.geometry import Polygon

from linz_stac_utils.elevation import load_elevation
```

```python
bbox = (172.6300, -43.5350, 172.6400, -43.5250)
```

```python
da = load_elevation(
    bbox=bbox,
    resolution=10,
    output_path="christchurch-dem.tif",
)

da
```

```python
polygon = Polygon(
    [
        (172.6300, -43.5350),
        (172.6400, -43.5350),
        (172.6350, -43.5250),
    ]
)

polygon_dem = load_elevation(intersects=polygon, resolution=10)
polygon_dem
```
